## 🧠 Additional Challenge: display a chart to summarize the entire data set

Read the entire data from the database and use it to plot a chart that summarizes the behavior of the robots based on their energy consumption.

#### Sample solution:

```python
# OWNER: Carlos
# Static view of a row slice from the full dataset.
# Contract: reads fetch_all()
# Edit ROW_START / ROW_END below to change which rows are plotted.
#
# NOTE: Rangeetha's DB cell above (get_connection/insert_reading/fetch_all) is still
# an unimplemented skeleton, so fetch_all isn't defined anywhere else in the notebook
# yet. Importing it directly from the already-built src/web-ui/db.py stub here so this
# cell runs standalone; swap this import for Rangeetha's real cell once it lands.

import sys
from pathlib import Path

import matplotlib.pyplot as plt

_WEB_UI_DIR = str(Path("src") / "web-ui")
if _WEB_UI_DIR not in sys.path:
    sys.path.insert(0, _WEB_UI_DIR)

from db import fetch_all

ROW_START = 0
ROW_END = 500


def plot_row_slice(df, row_start: int, row_end: int, output_path: str = None):
    """Static stacked-area chart of axis_1..axis_8 current over df.iloc[row_start:row_end],
    plus a stats callout box (peak axis, average active load)."""
    axis_cols = [f"axis_{i}" for i in range(1, 9)]
    sliced = df.iloc[row_start:row_end]

    plt.style.use("seaborn-v0_8-darkgrid")
    fig, ax = plt.subplots(figsize=(14, 6))

    ax.stackplot(
        sliced["reading_time"],
        *[sliced[c] for c in axis_cols],
        labels=[f"Axis {i}" for i in range(1, 9)],
    )
    ax.set_title(f"Robot Current Draw — Rows {row_start}:{row_end}")
    ax.set_xlabel("Time")
    ax.set_ylabel("Current (A)")
    ax.legend(loc="upper left", ncol=4, fontsize="small")

    active = sliced[sliced[axis_cols].sum(axis=1) > 0]
    peak_axis = sliced[axis_cols].max().idxmax()
    peak_value = sliced[axis_cols].max().max()
    stats_text = (
        f"Rows plotted: {len(sliced):,}\n"
        f"Active readings: {len(active):,} ({len(active) / len(sliced):.1%})\n"
        f"Peak axis: {peak_axis} ({peak_value:.1f} A)"
    )
    ax.text(
        0.99, 0.97, stats_text, transform=ax.transAxes,
        ha="right", va="top", fontsize=9,
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.85),
    )

    fig.tight_layout()
    if output_path:
        fig.savefig(output_path, dpi=150)
    return fig


full_df = fetch_all()
_ = plot_row_slice(full_df, ROW_START, ROW_END, output_path="summary_chart.png")
```
